# 🌐 Language Translation Tool

## Task 1 — Language Translation Tool

This project is a **Language Translation Tool** that allows users to enter text, select a source language and a target language, translate the text using a translation API, and listen to the input or translated text using Text-to-Speech (TTS).

### 🎯 Objectives

The application is designed to:

- Allow users to enter text through a simple interface.
- Allow users to select the **source** and **target** languages.
- Translate text using the **LibreTranslate API**.
- Display the translated text clearly.
- Provide a **copy** option for both the input and translated text.
- Provide **Text-to-Speech (TTS)** for supported languages.
- Allow generated audio to be downloaded.
- Clearly inform the user when TTS is unavailable for a selected language.
- Provide a clean, simple, and user-friendly interface.

### 🛠️ Technologies Used

- **Python** — Main programming language
- **Google Colab** — Development and execution environment
- **LibreTranslate** — Translation API/service
- **Gradio** — User interface
- **gTTS (Google Text-to-Speech)** — Text-to-Speech functionality
- **Requests** — Communication with the translation API

### 🔄 Application Workflow

```text

       ↓

````python
import gradio as gr
import requests
from gtts import gTTS
from gtts.lang import tts_langs
import tempfile


# =========================================================
# LANGUAGES
# =========================================================

language_choices = {
    lang["name"]: lang["code"]
    for lang in languages
}

language_names = sorted(language_choices.keys())


# =========================================================
# TTS LANGUAGES
# =========================================================

# Get all languages currently supported by gTTS
gtts_languages = tts_langs()

# Match LibreTranslate languages with gTTS languages
tts_languages = {}

for language_name, language_code in language_choices.items():

    if language_code in gtts_languages:
        tts_languages[language_name] = language_code


# Handle language-code differences between LibreTranslate and gTTS
tts_code_aliases = {
    "zh-Hans": "zh-CN",
    "zh-Hant": "zh-TW",
    "pt-BR": "pt",
}

for language_name, language_code in language_choices.items():

    if language_code in tts_code_aliases:

        gtts_code = tts_code_aliases[language_code]

        if gtts_code in gtts_languages:
            tts_languages[language_name] = gtts_code


# =========================================================
# TRANSLATION
# =========================================================

def translate_text(text, source_language, target_language):

    if not text or not text.strip():
        return ""

    response = requests.post(
        "http://localhost:5000/translate",
        data={
            "q": text,
            "source": language_choices[source_language],
            "target": language_choices[target_language]
        }
    )

    response.raise_for_status()

    return response.json()["translatedText"]


# =========================================================
# TEXT TO SPEECH
# =========================================================

def text_to_speech(text, language):

    if not text or not text.strip():
        return None, ""

    if language not in tts_languages:

        message = (
            f"⚠️ Audio for **{language}** cannot be generated "
            f"because text-to-speech is not available for this language."
        )

        return None, message

    filename = tempfile.NamedTemporaryFile(
        delete=False,
        suffix=".mp3"
    ).name

    gTTS(
        text=text,
        lang=tts_languages[language]
    ).save(filename)

    return filename, "✅ Audio generated successfully."


# =========================================================
# COPY
# =========================================================

def copy_text(text):
    return text


# =========================================================
# CSS
# =========================================================

custom_css = """

.gradio-container {
    max-width: 1100px !important;
    margin: auto !important;
    padding: 20px 25px !important;
}

.title {
    text-align: center;
    margin-bottom: 3px;
}

.subtitle {
    text-align: center;
    color: #6b7280;
    margin-bottom: 20px;
}

.translation-row {
    display: flex !important;
    flex-direction: row !important;
    gap: 20px !important;
}

.translation-column {
    flex: 1 !important;
    min-width: 0 !important;
}

.translation-column textarea {
    min-height: 180px !important;
    border-radius: 14px !important;
}

.action-row {
    margin-top: 8px !important;
}

.translate-button {
    margin-top: 15px !important;
}

"""


# =========================================================
# APP
# =========================================================

with gr.Blocks() as app:

    gr.Markdown(
        """
        <div class="title">
            <h1>🌐 Language Translation Tool</h1>
        </div>
        """
    )


    # -----------------------------------------------------
    # LANGUAGE DROPDOWNS
    # -----------------------------------------------------

    with gr.Row():

        source_language = gr.Dropdown(
            choices=language_names,
            value="English",
            label="From"
        )

        target_language = gr.Dropdown(
            choices=language_names,
            value="Urdu",
            label="To"
        )


    # -----------------------------------------------------
    # SIDE-BY-SIDE TEXT AREAS
    # -----------------------------------------------------

    with gr.Row(elem_classes="translation-row"):

        # LEFT SIDE
        with gr.Column(elem_classes="translation-column"):

            input_text = gr.Textbox(
                label="Your Text",
                placeholder="Enter text here...",
                lines=7,
                buttons=["copy"]
            )

            with gr.Row(elem_classes="action-row"):

                input_speak = gr.Button("🔊 Listen")


            input_audio = gr.Audio(
                label="Input Audio",
                type="filepath",
                buttons=["download"],
                visible=True
            )

            input_audio_status = gr.Markdown()


        # RIGHT SIDE
        with gr.Column(elem_classes="translation-column"):

            output_text = gr.Textbox(
                label="Translation",
                placeholder="Your translation will appear here...",
                lines=7,
                interactive=False,
                buttons=["copy"]
            )

            with gr.Row(elem_classes="action-row"):

                output_speak = gr.Button("🔊 Listen")


            output_audio = gr.Audio(
                label="Translation Audio",
                type="filepath",
                buttons=["download"],
                visible=True
            )

            output_audio_status = gr.Markdown()


    # -----------------------------------------------------
    # TRANSLATE BUTTON
    # -----------------------------------------------------

    translate_button = gr.Button(
        "✨ Translate",
        variant="primary",
        elem_classes="translate-button"
    )


    # -----------------------------------------------------
    # TRANSLATION ACTION
    # -----------------------------------------------------

    translate_button.click(
        fn=translate_text,
        inputs=[
            input_text,
            source_language,
            target_language
        ],
        outputs=output_text
    )


    # -----------------------------------------------------
    # INPUT TTS
    # -----------------------------------------------------

    input_speak.click(
        fn=text_to_speech,
        inputs=[
            input_text,
            source_language
        ],
        outputs=[
            input_audio,
            input_audio_status
        ]
    )


    # -----------------------------------------------------
    # OUTPUT TTS
    # -----------------------------------------------------

    output_speak.click(
        fn=text_to_speech,
        inputs=[
            output_text,
            target_language
        ],
        outputs=[
            output_audio,
            output_audio_status
        ]
    )


# =========================================================
# LAUNCH — GRADIO 6
# =========================================================

app.launch(
    theme=gr.themes.Soft(),
    css=custom_css
)
````

# 📝 Final Insights

## Project Summary

The Language Translation Tool successfully combines a translation API with a user-friendly web interface and Text-to-Speech functionality.

The final application allows users to:

- Enter text in the input area.
- Select from **50 available translation languages**.
- Translate text through the **LibreTranslate API**.
- View the original text and translation **side-by-side**.
- Copy text using the built-in copy buttons.
- Generate speech for supported languages.
- Download generated audio files.
- Receive a clear message when TTS is not available for a language.

## 🔊 Text-to-Speech Handling

Text-to-Speech support was implemented using **gTTS**.

Instead of assuming that every translation language can be converted to speech, the application checks whether the selected language is supported by gTTS.

For unsupported languages, the application displays a message such as:

> ⚠️ Audio for Persian cannot be generated because text-to-speech is not available for this language.

This allows translation to continue working normally even when audio generation is unavailable.
